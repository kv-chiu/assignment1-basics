### a

```python
compare_encodings = lambda s: {
    "original": s,
    "utf-8": {
        "bytes": s.encode("utf-8"),
        "len": len(s.encode("utf-8"))
    },
    "utf-16": {
        "bytes": s.encode("utf-16"),
        "len": len(s.encode("utf-16"))
    },
    "utf-32": {
        "bytes": s.encode("utf-32"),
        "len": len(s.encode("utf-32"))
    }
}

print("=== ASCII str ===")
print(compare_encodings("A"))
print("\n=== Chinese str ===")
print(compare_encodings("中"))
print("\n=== Emoji ===")
print(compare_encodings("😀"))
```

### b

```python
def decode_utf8_bytes_to_str_wrong(bytestring: bytes):
    return "".join([bytes([b]).decode("utf-8") for b in bytestring])
```

```python
decode_utf8_bytes_to_str_wrong("hello".encode("utf-8"))
```

```python
decode_utf8_bytes_to_str_wrong("A中😀".encode("utf-8"))
```
